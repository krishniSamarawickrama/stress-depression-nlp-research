```python

import pandas as pd
import re

# Create a separate copy for NLP
nlp_df = df.copy()

# Identify the four text-response columns
text_cols = [
    col for col in nlp_df.columns
    if any(term in col.lower() for term in [
        "how do you usually feel",
        "what kinds of situations",
        "how do you describe your feelings",
        "anything else"
    ])
]

def clean_text(text):
    if pd.isna(text):
        return ""

    text = str(text).lower().strip()
    text = re.sub(r'\s+', ' ', text)
    text = re.sub(r'[^\w\s.,!?\'-]', '', text)
    return text

for col in text_cols:
    nlp_df[col + "_clean"] = nlp_df[col].apply(clean_text)

print("Text cleaning completed.")
print("Number of text columns:", len(text_cols))

for col in text_cols:
    clean_col = col + "_clean"
    print("\nQuestion:", col.strip())
    print("Non-empty responses:", (nlp_df[clean_col] != "").sum())
    print("Unique responses:", nlp_df.loc[
        nlp_df[clean_col] != "", clean_col
    ].nunique())
```

```python

for col in text_cols:
    clean_col = col + "_clean"
    answers = nlp_df[clean_col]
    valid_answers = answers[answers != ""]

    short_count = valid_answers.apply(
        lambda x: len(x.split()) <= 2
    ).sum()

    long_count = valid_answers.apply(
        lambda x: len(x.split()) > 2
    ).sum()

    print("\nQuestion:", col.strip())
    print("Short answers (2 words or fewer):", short_count)
    print("Longer answers (more than 2 words):", long_count)
```

```python

import pandas as pd
import re

# Use only consented responses
text_columns = list(analysis_df.columns[11:15])

records = []

for participant_no, (_, row) in enumerate(analysis_df.iterrows(), start=1):
    for question_no, column in enumerate(text_columns, start=1):
        answer = row[column]

        if pd.notna(answer) and str(answer).strip():
            original_text = str(answer).strip()

            # Basic text preprocessing
            clean_text = original_text.lower()
            clean_text = re.sub(r"https?://\S+|www\.\S+", " ", clean_text)
            clean_text = re.sub(r"[^a-zA-Z0-9\s']", " ", clean_text)
            clean_text = re.sub(r"\s+", " ", clean_text).strip()

            if clean_text:
                records.append({
                    "participant_id": participant_no,
                    "question_no": question_no,
                    "text": original_text,
                    "clean_text": clean_text
                })

preprocessed_df = pd.DataFrame(records)

print("Preprocessing completed")
print("Participants:", analysis_df.shape[0])
print("Non-empty text answers:", len(preprocessed_df))
display(preprocessed_df.head(10))

preprocessed_df.to_csv(
    "preprocessed_text_responses.csv",
    index=False
)

print("Saved: preprocessed_text_responses.csv")
```

```python

import pandas as pd
import numpy as np

from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support,
    classification_report,
    confusion_matrix
)

# Use the emotion dataset loaded in the previous step
train_df = dataset["train"].to_pandas()
test_df = dataset["test"].to_pandas()

X_train = train_df["text"].astype(str)
y_train = train_df["label"]

X_test = test_df["text"].astype(str)
y_test = test_df["label"]

# Baseline model
baseline_model = Pipeline([
    ("tfidf", TfidfVectorizer(
        lowercase=True,
        ngram_range=(1, 2),
        max_features=20000
    )),
    ("classifier", LogisticRegression(
        max_iter=1000,
        random_state=42
    ))
])

# Train
baseline_model.fit(X_train, y_train)

# Test on the separate test set
y_pred = baseline_model.predict(X_test)

# Evaluate
accuracy = accuracy_score(y_test, y_pred)
precision, recall, macro_f1, _ = precision_recall_fscore_support(
    y_test, y_pred, average="macro", zero_division=0
)

print("BASELINE MODEL: TF-IDF + LOGISTIC REGRESSION")
print(f"Accuracy:  {accuracy:.4f}")
print(f"Macro Precision: {precision:.4f}")
print(f"Macro Recall:    {recall:.4f}")
print(f"Macro F1-score:  {macro_f1:.4f}")

print("\nClassification Report:")
print(classification_report(y_test, y_pred, zero_division=0))
```

```python

from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC
from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support,
    classification_report
)

# SVM comparison model
svm_model = Pipeline([
    ("tfidf", TfidfVectorizer(
        lowercase=True,
        ngram_range=(1, 2),
        max_features=20000
    )),
    ("classifier", LinearSVC(random_state=42))
])

# Train on the training split
svm_model.fit(X_train, y_train)

# Evaluate on the separate test split
svm_pred = svm_model.predict(X_test)

accuracy = accuracy_score(y_test, svm_pred)

precision, recall, macro_f1, _ = precision_recall_fscore_support(
    y_test,
    svm_pred,
    average="macro",
    zero_division=0
)

print("SVM MODEL: TF-IDF + Linear SVM")
print(f"Accuracy:       {accuracy:.4f}")
print(f"Macro Precision: {precision:.4f}")
print(f"Macro Recall:    {recall:.4f}")
print(f"Macro F1-score:  {macro_f1:.4f}")

print("\nClassification Report:")
print(classification_report(y_test, svm_pred, zero_division=0))
```

```python

from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
import pandas as pd

# Use training split only for 5-fold cross-validation
X_cv = X_train.reset_index(drop=True)
y_cv = y_train.reset_index(drop=True)

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

models = {
    "Logistic Regression": Pipeline([
        ("tfidf", TfidfVectorizer(
            ngram_range=(1, 2),
            max_features=20000
        )),
        ("classifier", LogisticRegression(
            max_iter=1000,
            random_state=42
        ))
    ]),
    "Linear SVM": Pipeline([
        ("tfidf", TfidfVectorizer(
            ngram_range=(1, 2),
            max_features=20000
        )),
        ("classifier", LinearSVC(random_state=42))
    ])
}

scoring = {
    "accuracy": "accuracy",
    "precision_macro": "precision_macro",
    "recall_macro": "recall_macro",
    "f1_macro": "f1_macro"
}

results = []

for model_name, model in models.items():
    scores = cross_validate(
        model,
        X_cv,
        y_cv,
        cv=cv,
        scoring=scoring,
        n_jobs=-1
    )

    row = {"Model": model_name}

    for metric in scoring:
        values = scores["test_" + metric]
        row[metric + "_mean"] = values.mean()
        row[metric + "_std"] = values.std(ddof=1)

    results.append(row)

cv_results = pd.DataFrame(results)

display(cv_results.round(4))

cv_results.to_csv(
    "five_fold_cross_validation_results.csv",
    index=False
)

print("Saved: five_fold_cross_validation_results.csv")
```

```python

import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay, confusion_matrix

# Predict using the SVM model on the separate test set
svm_pred = svm_model.predict(X_test)

cm = confusion_matrix(y_test, svm_pred)

labels = dataset["train"].features["label"].names

fig, ax = plt.subplots(figsize=(8, 6))

display_matrix = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=labels
)

display_matrix.plot(
    ax=ax,
    cmap="Blues",
    xticks_rotation=45,
    colorbar=False
)

plt.title("Linear SVM - Emotion Classification Confusion Matrix")
plt.tight_layout()
plt.savefig("svm_confusion_matrix.png", dpi=300, bbox_inches="tight")
plt.show()

print("Saved: svm_confusion_matrix.png")
```

```python

# Identify incorrectly classified emotion examples

failure_df = pd.DataFrame({
    "text": X_test.reset_index(drop=True),
    "actual_label": y_test.reset_index(drop=True),
    "predicted_label": svm_pred
})

failure_df = failure_df[
    failure_df["actual_label"] != failure_df["predicted_label"]
].copy()

label_names = dataset["train"].features["label"].names

failure_df["actual_emotion"] = failure_df["actual_label"].apply(
    lambda x: label_names[x]
)
failure_df["predicted_emotion"] = failure_df["predicted_label"].apply(
    lambda x: label_names[x]
)

print("Total incorrect predictions:", len(failure_df))
print("\nCommon error types:")

display(
    failure_df.groupby(
        ["actual_emotion", "predicted_emotion"]
    ).size().reset_index(name="count")
    .sort_values("count", ascending=False)
    .head(10)
)

print("\nExample misclassifications:")
display(failure_df.head(10))

failure_df.to_csv("svm_failure_analysis.csv", index=False)
print("Saved: svm_failure_analysis.csv")
```
